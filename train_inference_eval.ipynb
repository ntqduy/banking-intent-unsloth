```python
# Compare performance using evaluation outputs
import pandas as pd
from pathlib import Path

eval_csv = Path('result/evaluate_both/compare_base_finetune.csv')
inf_csv = Path('result/inf_both/compare_base_finetune.csv')

if eval_csv.exists():
    print('Evaluation comparison:')
    display(pd.read_csv(eval_csv))
else:
    print(f'Missing file: {eval_csv}. Run: !bash evaluate.sh both')

if inf_csv.exists():
    print('Inference comparison:')
    display(pd.read_csv(inf_csv))
else:
    print(f'Missing file: {inf_csv}. Run: !bash inference.sh both')
```
